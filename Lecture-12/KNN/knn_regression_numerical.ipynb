```python
import numpy as np
import matplotlib.pyplot as plt
# Dataset: x and corresponding y (target)
X = np.array([1, 2, 3, 4, 5])
y = np.array([1.5, 1.7, 3.0, 3.2, 5.0])
# Query point
x_query = 2.5
# Step 1: Calculate distances (Euclidean or Manhattan, same for 1D)
distances = np.abs(X - x_query)
# Step 2: Find indices of k=3 nearest neighbors
k = 3
nearest_indices = distances.argsort()[:k]
# Step 3: Get their y values
nearest_y = y[nearest_indices]
# Step 4: Predict as mean of y values
y_pred = nearest_y.mean()
print(f"Nearest x values: {X[nearest_indices]}")
print(f"Corresponding y values: {nearest_y}")
print(f"Predicted value at x = {x_query} is y = {y_pred:.2f}")
# --- Plotting ---
plt.figure(figsize=(7, 5))
# Plot all data points
plt.scatter(X, y, color='blue', label='Training Points')
for i, txt in enumerate(y):
   plt.annotate(f"P{i+1}", (X[i] + 0.05, y[i] + 0.05))
# Highlight the query point
plt.scatter(x_query, y_pred, color='red', label=f'Prediction at x={x_query}', marker='X', s=100)
# Draw lines to neighbors
for idx in nearest_indices:
   plt.plot([X[idx], x_query], [y[idx], y_pred], 'k--', alpha=0.6)
plt.axvline(x=x_query, color='gray', linestyle=':', label='Query x')
plt.title("KNN Regression (k=3)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
'''Output -
Nearest x values: [2 3 1]
Corresponding y values: [1.7 3.  1.5]
Predicted value at x = 2.5 is y = 2.07
'''
```
