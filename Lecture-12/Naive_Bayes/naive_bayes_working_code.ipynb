```python
from collections import Counter, defaultdict
# Dataset
data = [
    {'Study': 'Yes', 'Sleep': 'Enough',     'Result': 'Pass'},
    {'Study': 'Yes', 'Sleep': 'Not enough', 'Result': 'Pass'},
    {'Study': 'No',  'Sleep': 'Enough',     'Result': 'Fail'},
    {'Study': 'No',  'Sleep': 'Not enough', 'Result': 'Fail'},
    {'Study': 'Yes', 'Sleep': 'Enough',     'Result': 'Pass'},
]
# Input to predict
query = {'Study': 'Yes', 'Sleep': 'Not enough'}
# Step 1: Calculate Prior Probabilities
label_counts = Counter([row['Result'] for row in data])
total = len(data)
priors = {label: count / total for label, count in label_counts.items()}
# Step 2: Calculate Likelihoods
# Likelihoods: P(feature_value | class)
likelihoods = defaultdict(lambda: defaultdict(int))
# Count feature values for each class
feature_counts = defaultdict(lambda: defaultdict(Counter))
for row in data:
    label = row['Result']
    for feature in ['Study', 'Sleep']:
        value = row[feature]
        feature_counts[label][feature][value] += 1
# Convert counts to probabilities
for label in label_counts:
    for feature in ['Study', 'Sleep']:
        total_label = label_counts[label]
        for value, count in feature_counts[label][feature].items():
            likelihoods[label][(feature, value)] = count / total_label
# Step 3: Compute Posterior for each class
posteriors = {}
for label in label_counts:
    posterior = priors[label]
    for feature in query:
        value = query[feature]
        posterior *= likelihoods[label].get((feature, value), 0)  # 0 if not seen
    posteriors[label] = posterior
# Step 4: Prediction
prediction = max(posteriors, key=posteriors.get)
# Output
print("Posterior probabilities:")
for label in posteriors:
    print(f"{label}: {posteriors[label]:.4f}")
print(f"\n✅ Predicted class: {prediction}")
'''Output -
Posterior probabilities:
Pass: 0.2000
Fail: 0.0000
✅ Predicted class: Pass
'''
```
