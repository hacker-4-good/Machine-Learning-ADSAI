```python
import numpy as np
import math
# Dataset
data = [
    (1.4, 0),  # Iris-setosa
    (1.3, 0),
    (1.5, 0),
    (4.5, 1),  # Iris-versicolor
    (4.7, 1),
    (4.6, 1),
]
# Separate by class
class0 = [x for x, y in data if y == 0]
class1 = [x for x, y in data if y == 1]
# Mean and variance
mu_0 = np.mean(class0)
sigma2_0 = np.var(class0, ddof=0)
mu_1 = np.mean(class1)
sigma2_1 = np.var(class1, ddof=0)
# Gaussian PDF
def gaussian_pdf(x, mu, sigma2):
    return (1 / math.sqrt(2 * math.pi * sigma2)) * math.exp(-(x - mu) ** 2 / (2 * sigma2))
# New sample
x_new = 1.6
# Likelihoods
likelihood_0 = gaussian_pdf(x_new, mu_0, sigma2_0)
likelihood_1 = gaussian_pdf(x_new, mu_1, sigma2_1)
# Priors (equal)
prior_0 = prior_1 = 0.5
# Posterior (unnormalized)
posterior_0 = likelihood_0 * prior_0
posterior_1 = likelihood_1 * prior_1
# Prediction
predicted_class = 0 if posterior_0 > posterior_1 else 1
# Output
print(f"Class 0 mean = {mu_0}, variance = {sigma2_0}")
print(f"Class 1 mean = {mu_1}, variance = {sigma2_1}")
print(f"\nP(x=1.6 | C=0) ≈ {likelihood_0:.4f}")
print(f"P(x=1.6 | C=1) ≈ {likelihood_1:.4f}")
print(f"\nPosterior C=0: {posterior_0:.4f}")
print(f"Posterior C=1: {posterior_1:.4f}")
print(f"\n✅ Predicted class: {predicted_class} ({'Iris-setosa' if predicted_class == 0 else 'Iris-versicolor'})")
'''Output -
Class 0 mean = 1.4000000000000001, variance = 0.006666666666666664
Class 1 mean = 4.6, variance = 0.006666666666666678
P(x=1.6 | C=0) ≈ 0.2433
P(x=1.6 | C=1) ≈ 0.0000
Posterior C=0: 0.1216
Posterior C=1: 0.0000
✅ Predicted class: 0 (Iris-setosa)
'''
```
