```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# Create the dataset
data = {
   'Gender': ['M', 'M', 'F', 'M', 'F', 'M', 'M'],
   'Age': [48, 67, 53, 49, 32, 34, 21],
   'Temp': [102.2, 101.5, 98.6, 101.4, 98.7,99.1,98.4],
   'Label': ['sick', 'sick', 'healthy', 'sick', 'healthy', 'healthy', 'healthy']
}
df = pd.DataFrame(data)
# Plotting
plt.figure(figsize = (8,3))
sns.scatterplot(x='Gender', y='Age', hue='Label', data=df, palette={'sick': 'red', 'healthy': 'green'}, s=100)
plt.title('Age vs Gender (colored by Health Label)')
plt.ylabel('Age')
plt.grid(True)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
df = pd.DataFrame(data)
# Plotting
plt.figure(figsize = (8,3))
sns.scatterplot(x='Age', y='Temp', hue='Label', data=df, palette={'sick': 'red', 'healthy': 'green'}, s=100)
plt.title('Age vs Temp (colored by Health Label)')
plt.ylabel('Age')
plt.grid(True)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
