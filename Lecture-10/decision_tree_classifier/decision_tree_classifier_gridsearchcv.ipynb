```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV
import pandas as pd
# Load data
X, y = load_iris(return_X_y=True)
# Define the model
dtree = DecisionTreeClassifier(random_state=42)
# Define hyperparameter grid
param_grid = {
   'criterion': ['gini', 'entropy', 'log_loss'],
   'max_depth': [2, 3, 4, 5, None],
   'min_samples_split': [2, 4, 6],
   'min_samples_leaf': [1, 2, 3]
}
# Setup GridSearchCV
grid_search = GridSearchCV(
   estimator=dtree,
   param_grid=param_grid,
   cv=5,                 # 5-fold cross-validation
   scoring='accuracy',   # Metric to optimize
   n_jobs=-1             # Use all CPU cores
)
# Run grid search
grid_search.fit(X, y)
# Show best parameters and score
print("✅ Best Parameters:")
print(grid_search.best_params_)
print(f"\n🎯 Best Cross-Validation Accuracy: {grid_search.best_score_:.4f}")
# Convert to DataFrame to explore all results
results_df = pd.DataFrame(grid_search.cv_results_)
print("\nTop 5 Configurations:")
print(results_df[['mean_test_score', 'params']].sort_values(by='mean_test_score', ascending=False).head())
```
