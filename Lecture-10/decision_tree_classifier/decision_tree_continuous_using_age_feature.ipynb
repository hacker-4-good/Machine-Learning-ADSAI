```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
df3 = pd.DataFrame({
   'Age': [25, 45, 35, 33, 52, 23],
   'Income': [50, 80, 60, 85, 45, 77],
   'Buy': [0, 1, 0, 1, 0, 1]
})
X3 = df3[['Income']]
y3 = df3['Buy']
model3 = DecisionTreeClassifier(criterion='gini', max_depth=5)
model3.fit(X3, y3)
plt.figure(figsize=(15, 6))
plot_tree(model3, feature_names=X3.columns, class_names=['No', 'Yes'], filled=True)
plt.show()
```
