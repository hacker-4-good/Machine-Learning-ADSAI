```python
import pandas as pd
# Dataset
data = {
    'Age': [25, 45, 35, 33, 52, 23],
    'Income': [50, 80, 60, 85, 45, 77],
    'Buy': [0, 1, 0, 1, 0, 1]
}
df = pd.DataFrame(data)
# Gini for label list
def gini(labels):
    probs = pd.Series(labels).value_counts(normalize=True)
    return 1 - sum(probs**2)
# Compute Gini for each threshold
def gini_for_feature(df, feature, target):
    df_sorted = df.sort_values(feature).reset_index(drop=True)
    values = df_sorted[feature].unique()
    midpoints = [(values[i] + values[i+1]) / 2 for i in range(len(values)-1)]
    print(f"\nGini values for feature: {feature}")
    for threshold in midpoints:
        left = df[df[feature] <= threshold][target]
        right = df[df[feature] > threshold][target]
        gini_left = gini(left)
        gini_right = gini(right)
        weighted_gini = (len(left) * gini_left + len(right) * gini_right) / len(df)
        print(f"Threshold {threshold:.2f} → Gini = {weighted_gini:.4f}")
# Apply for both features
gini_for_feature(df, 'Age', 'Buy')
gini_for_feature(df, 'Income', 'Buy')
```
