```python
import pandas as pd
# Create the dataset
data = {
    'Weather': ['Sunny', 'Sunny', 'Rainy', 'Rainy', 'Rainy', 'Sunny'],
    'Wind': ['Weak', 'Strong', 'Weak', 'Strong', 'Weak', 'Weak'],
    'Play': ['No', 'No', 'Yes', 'No', 'Yes', 'Yes']
}
df = pd.DataFrame(data)
# Function to compute Gini Index for a column
def gini_index(column, target):
    groups = column.unique()
    gini = 0.0
    for group in groups:
        subset = df[column == group]
        size = len(subset)
        if size == 0:
            continue
        score = 0.0
        for label in target.unique():
            p = len(subset[subset == label]) / size
            score += p ** 2
        gini += (1 - score) * (size / len(df))
    return gini
# Gini for the root node
def gini_root(df, target_col):
    target_counts = df[target_col].value_counts(normalize=True)
    gini = 1 - sum(p ** 2 for p in target_counts)
    return gini
# Example usage
print("Gini at root:", gini_root(df, 'Wind'))
print("Gini at Weather:", gini_index(df, 'Weather'))
```
