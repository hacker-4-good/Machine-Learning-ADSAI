```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
# Load dataset
X, y = load_iris(return_X_y=True)
import pandas as pd 
data = load_iris()
iris_data = pd.DataFrame(data.data, columns=data.feature_names)
print(iris_data.head())
# Train Decision Tree with multiple hyperparameters
model = DecisionTreeClassifier(
   # pass different criterion (log_loss & gini)
   criterion='entropy',        # Use entropy to split 
   max_depth=2,                # Limit depth to 3
   min_samples_split=4,        # At least 4 samples to split
   min_samples_leaf=2,         # Each leaf must have ≥2 samples
   max_features=None,          # Use all features
   random_state=42
)
model.fit(X, y)
# Visualize the tree
plt.figure(figsize=(15, 8))
plot_tree(model, filled=True, feature_names=load_iris().feature_names, class_names=load_iris().target_names)
plt.title("Decision Tree with Custom Hyperparameters")
plt.show()
```
