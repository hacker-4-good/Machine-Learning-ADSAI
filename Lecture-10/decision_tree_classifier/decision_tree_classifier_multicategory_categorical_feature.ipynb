```python
import pandas as pd
# Create the dataset
data = {
    'Color': ['Red', 'Green', 'Blue', 'Red', 'Blue', 'Green'],
    'Size': ['Small', 'Medium', 'Large', 'Medium', 'Small', 'Large'],
    'Label': ['A', 'A', 'B', 'B', 'A', 'B']
}
df = pd.DataFrame(data)
# Gini index at the root
def gini_root(df, target_col='Label'):
    probs = df[target_col].value_counts(normalize=True)
    gini = 1 - sum(p**2 for p in probs)
    return gini
# Gini after split by a column (for all values in the column)
def gini_split(df, feature_col, target_col='Label'):
    groups = df[feature_col].unique()
    gini = 0.0
    for group in groups:
        subset = df[df[feature_col] == group]
        size = len(subset)
        if size == 0:
            continue
        probs = subset[target_col].value_counts(normalize=True)
        score = 1 - sum(p**2 for p in probs)
        gini += (size / len(df)) * score
    return gini
# Example usage
print("Gini at root:", gini_root(df))
print("Gini after splitting by 'Color':", gini_split(df, 'Color'))
print("Gini after splitting by 'Size':", gini_split(df, 'Size'))
```
