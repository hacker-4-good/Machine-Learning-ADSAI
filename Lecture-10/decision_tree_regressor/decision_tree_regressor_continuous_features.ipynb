```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor, plot_tree
import matplotlib.pyplot as plt
df3 = pd.DataFrame({
    'Area': [1000, 800, 1200, 1500, 900, 1100, 1300, 1400, 850, 950],
    'Bedrooms': [2, 2, 3, 4, 1, 3, 3, 4, 1, 2],
    'Rent': [2000, 1800, 2500, 3000, 1600, 2300, 2700, 2900, 1700, 1900]
})
X3 = df3[['Area', 'Bedrooms']]
y3 = df3['Rent']
model3 = DecisionTreeRegressor(max_depth=3)
model3.fit(X3, y3)
plt.figure(figsize=(15, 5))
plot_tree(model3, feature_names=X3.columns, filled=True)
plt.show()
```
