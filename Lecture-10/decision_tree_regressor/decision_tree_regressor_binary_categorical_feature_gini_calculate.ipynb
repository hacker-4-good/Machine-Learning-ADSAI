```python
import pandas as pd
# Create DataFrame
data = {
    'Feature1': ['A', 'A', 'A', 'A', 'B', 'B', 'B', 'B'],
    'Feature2': ['X', 'X', 'Y', 'Y', 'X', 'X', 'Y', 'Y'],
    'Target': [10, 12, 18, 22, 30, 28, 35, 40]
}
df = pd.DataFrame(data)
# Function to compute SSE for a given categorical split
def sse_split(df, feature, target):
    sse_total = 0
    groups = df[feature].unique()
    for group in groups:
        subset = df[df[feature] == group]
        mean_val = subset[target].mean()
        sse = ((subset[target] - mean_val) ** 2).sum()
        print(f"Group {group} → Mean: {mean_val:.2f}, SSE: {sse:.2f}")
        sse_total += sse
    print(f"\nTotal SSE for split on {feature}: {sse_total:.2f}")
    return sse_total
# Example usage
sse_split(df, 'Feature1', 'Target')
sse_split(df, 'Feature2', 'Target')
```
