```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor, plot_tree
import matplotlib.pyplot as plt
df1 = pd.DataFrame({
    'Feature1': ['A']*4 + ['B']*4,
    'Feature2': ['X','X','Y','Y','X','X','Y','Y'],
    'Target': [10,12,18,22,30,28,35,40]
})
X1 = pd.get_dummies(df1[['Feature1', 'Feature2']])
y1 = df1['Target']
model1 = DecisionTreeRegressor(max_depth=2)
model1.fit(X1, y1)
plt.figure(figsize=(10, 5))
plot_tree(model1, feature_names=X1.columns, filled=True)
plt.show()
```
