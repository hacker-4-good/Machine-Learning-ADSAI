```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor, plot_tree
import matplotlib.pyplot as plt
df2 = pd.DataFrame({
    'Color': ['Red', 'Blue', 'Green', 'Blue', 'Green', 'Red', 'Blue', 'Green', 'Red', 'Blue'],
    'Size': ['Small', 'Medium', 'Large', 'Large', 'Small', 'Medium', 'Small', 'Medium', 'Large', 'Medium'],
    'Price': [5, 6, 10, 9, 7, 4, 8, 11, 6, 10]
})
X2 = pd.get_dummies(df2[['Color', 'Size']])
y2 = df2['Price']
model2 = DecisionTreeRegressor(max_depth=4)
model2.fit(X2, y2)
plt.figure(figsize=(10, 5))
plot_tree(model2, feature_names=X2.columns, filled=True)
plt.show()
```
