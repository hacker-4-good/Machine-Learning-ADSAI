```python
import pandas as pd
# Create the dataset
data = {
    'Color': ['Red', 'Blue', 'Green', 'Blue', 'Green', 'Red', 'Blue', 'Green', 'Red', 'Blue'],
    'Size': ['Small', 'Medium', 'Large', 'Large', 'Small', 'Medium', 'Small', 'Medium', 'Large', 'Medium'],
    'Price': [5, 6, 10, 9, 7, 4, 8, 11, 6, 10]
}
df = pd.DataFrame(data)
# Compute SSE for split on a specific category
def sse_split_on_category(df, feature, value, target):
    group1 = df[df[feature] == value]
    group2 = df[df[feature] != value]
    mean1 = group1[target].mean()
    mean2 = group2[target].mean()
    sse1 = ((group1[target] - mean1) ** 2).sum()
    sse2 = ((group2[target] - mean2) ** 2).sum()
    total_sse = sse1 + sse2
    print(f"Group {value}: Mean = {mean1:.2f}, SSE = {sse1:.2f}")
    print(f"Group not {value}: Mean = {mean2:.2f}, SSE = {sse2:.2f}")
    print(f"\nTotal SSE = {total_sse:.2f}")
    return total_sse
# Example: Split on Color
print("Split of basis of Colors")
sse_split_on_category(df, 'Color', 'Blue', 'Price')
sse_split_on_category(df, 'Color', 'Green', 'Price')
sse_split_on_category(df, 'Color', 'Red', 'Price')
print("-"*50)
print("Split of basis of Colors")
sse_split_on_category(df, 'Size', 'Small', 'Price')
sse_split_on_category(df, 'Size', 'Medium', 'Price')
sse_split_on_category(df, 'Size', 'Large', 'Price')
```
