```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
# Dataset: [x1, x2] and labels y
X = np.array([
    [1, 2],  # A
   
    [6, 5],  # D
   
])
y = np.array([-1, 1,])
# Train SVM classifier
svm_clf = SVC(kernel='linear', C=1.0)
svm_clf.fit(X, y)
# Plotting
def plot_svm_decision_boundary(clf, X, y):
    plt.figure(figsize=(8, 6))
    # Plot original points
    plt.scatter(X[y==-1, 0], X[y==-1, 1], color='red', label='Class -1')
    plt.scatter(X[y==1, 0], X[y==1, 1], color='blue', label='Class +1')
    # Create grid to evaluate model
    ax = plt.gca()
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()
    xx = np.linspace(xlim[0]-1, xlim[1]+1, 100)
    yy = np.linspace(ylim[0]-1, ylim[1]+1, 100)
    YY, XX = np.meshgrid(yy, xx)
    xy = np.vstack([XX.ravel(), YY.ravel()]).T
    Z = clf.decision_function(xy).reshape(XX.shape)
    # Plot decision boundary and margins
    plt.contour(XX, YY, Z, levels=[-1, 0, 1], linestyles=['--', '-', '--'], colors='black')
    plt.scatter(clf.support_vectors_[:, 0], clf.support_vectors_[:, 1], s=100, facecolors='none', edgecolors='k', label='Support Vectors')
    plt.legend()
    plt.title("SVM Classification with Decision Boundary")
    plt.xlabel("x₁")
    plt.ylabel("x₂")
    plt.grid(True)
    plt.show()
# Call the plot function
plot_svm_decision_boundary(svm_clf, X, y)
```
